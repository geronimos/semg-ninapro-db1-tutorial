## NinaPro DB1 at a glance

**NinaPro** (Non-Invasive Neurally-mediated Interfaces for Prosthetics and Orthotics) is a public database of electromyography and hand gesture recordings for myoelectric research.

**DB1** is the original, smallest dataset in the family:

- **Subjects:** 27 intact subjects
- **Gestures:** 52 movements plus rest (53 classes)
- **Repetitions:** 10 per movement
- **Exercises:** 3 blocks (A1, A2, A3)
- **Modalities:** sEMG (10 channels) + CyberGlove (22 channels)
- **Sampling:** 100 Hz, synchronized

### File variables (per subject, per exercise)

| Variable | Shape | Description |
|----------|-------|-------------|
| `subject` | 1 | Subject ID (1-27) |
| `exercise` | 1 | Exercise ID (1-3) |
| `emg` | T × 10 | sEMG signals |
| `glove` | T × 22 | CyberGlove signals |
| `stimulus` | T | Original movement labels (1-52, 0=rest) |
| `restimulus` | T | Refined movement labels (recommended) |
| `repetition` | T | Repetition index of `stimulus` (1-10) |
| `rerepetition` | T | Repetition index of `restimulus` (1-10) |

**Recommendation:** use `restimulus` and `rerepetition` for analysis because the labels are temporally refined.

### Acquisition setup (sEMG)

- Channels 1-8 are equally spaced around the forearm.
- Channel 9 targets the Flexor Digitorum Superficialis.
- Channel 10 targets the Extensor Digitorum Superficialis.


## What DB1 can and cannot answer

**Good for:**
- Learning the EMG-to-gesture mapping problem.
- Studying windowing and labeling choices.
- Comparing within-subject vs. cross-subject generalization.
- Reproducing classic baselines from the literature.

**Not good for:**
- Real-world prosthetic control (able-bodied subjects, controlled lab conditions).
- Continuous, natural hand use (discrete gestures only).
- Spatial EMG modeling at high density (only 10 channels).
